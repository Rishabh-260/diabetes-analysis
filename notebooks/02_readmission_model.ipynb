## Interpretation
ROC-AUC summarizes ranking across thresholds, while precision and recall describe positive predictions at the fixed 0.5 threshold. F1 balances precision/recall. PR-AUC summarizes the precision-recall curve. Confusion matrices are saved under reports/model. Feature coefficients/importances are predictive associations, not causal effects. A deployment would require temporal and external validation, calibration, subgroup review, and clinical/operational input.
